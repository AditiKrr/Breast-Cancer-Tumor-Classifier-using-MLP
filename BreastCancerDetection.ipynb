Breast Cancer Classification with a simple Neural Network (tabular data)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

we try to see if the tumor is benign(1) or malignant(0): (malignant means cancerous)

```python
print(breast_cancer_dataset)
```

```python
df=pd.DataFrame(breast_cancer_dataset.data, columns=breast_cancer_dataset.feature_names)
```

```python
df.head()
```

```python
df['label']=breast_cancer_dataset.target
```

```python
df.tail()
```

```python
df.shape
```

```python
df.info()
```

no missing values in any of the columns

```python
df.describe()
```

```python
df['label'].value_counts()
```

```python
df.groupby('label').mean()
```

```python
X=df.drop(columns='label',axis=1)
Y=df['label']
```

```python
X.head()
```

```python
Y.head()
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=3)
```

```python
X_train.tail()
```

Scaling teh data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler =StandardScaler()
X_train_std=scaler.fit_transform(X_train)
X_test_std=scaler.transform(X_test)
```

```python
print(X_train_std)
```

# Building the Neural Network

```python
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
#setting up the layers for my neural network
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20, activation='relu'), #2 hidden layers
    keras.layers.Dense(20, activation='relu'),
    keras.layers.Dense(2, activation='softmax')
]) #2 because output can be 2 classes: 0(malignant) or 1(benign)
```

Compiling the **Neural** Network

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
#training the neural network
history=model.fit(X_train_std,Y_train, validation_split = 0.1, epochs=10) #within the train datset we test using validation set
```

VISUALISING THE ACCURACY AND THE LOSS

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epochs')
plt.legend(['training data','validation data'], loc= 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epochs')
plt.legend(['training data','validation data'], loc= 'lower right')
```

ACCURACY ON TEST DATA

```python
loss, accuracy = model.evaluate(X_test_std, Y_test)
```

CLASSIFICATION REPORT

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_std)

```

```python
print(y_pred.shape)
```

```python
print(y_pred[0]) #gives the probability of being 0(malignant) and probability of entry being 1
```

```python
print(y_pred)
```

```python
#converting the probabilities into labels (0:malignant, 1:benign)
#np.argmax() returns the index of the largest element in a list
#so if the first value is maximum (has a higher probability) we give label 0 or else we give label as 1 for that datapoint
y_pred_labels=[np.argmax(i) for i in y_pred]
print(y_pred_labels)

```

# Building the Predictive System

```python
input_data = (11.76,21.6,74.72,427.9,0.08637,0.04966,0.01657,0.01115,0.1495,0.05888,0.4062,1.21,2.635,28.47,0.005857,0.009758,0.01168,0.007445,0.02406,0.001769,12.98,25.72,82.98,516.5,0.1085,0.08615,0.05523,0.03715,0.2433,0.06563)

# change the input_data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the numpy array as we are predicting for one data point
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

# standardizing the input data
input_data_std = scaler.transform(input_data_reshaped)

prediction = model.predict(input_data_std)
print(prediction)

prediction_label = [np.argmax(prediction)]
print(prediction_label)

if(prediction_label[0] == 0):
  print('The tumor is Malignant')

else:
  print('The tumor is Benign')
```
